# EMPLOYEE ATTRITION PREDICTION USING MACHINE LEARNING

## 1. Introduction

Employee attrition is one of the major challenges faced by
organisations. It occurs when employees leave an organisation
voluntarily or involuntarily.

High employee turnover can result in increased recruitment
costs, reduced productivity, loss of organisational knowledge,
and additional expenses associated with training new employees.

Understanding the factors that contribute to employee attrition
can help organisations develop better employee retention
strategies.

This project focuses on developing a machine learning model
that predicts whether an employee is likely to leave an
organisation based on available employee information.

The project uses the IBM HR Analytics Employee Attrition
dataset.The dataset is fictional data created by IBM data scientists, not actual employee records from IBM's workforce.

## 2. Problem Statement

Organisations often experience difficulties identifying
employees who are at risk of leaving.

Without an effective analytical approach, management may
struggle to identify patterns associated with employee
turnover.

This project addresses the problem by applying machine
learning techniques to historical employee data to predict
employee attrition.

# 8. Data Quality Assessment

Before performing exploratory data analysis, the dataset
must be examined for potential data quality issues.

This section investigates:

1. Missing values.
2. Duplicate records.
3. Data types.
4. Unique values.
5. Constant columns.

These checks help determine whether the dataset requires
additional cleaning before model development.

```python
# Check missing values in each column

missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
# Check duplicate records

duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)
```

```python
# Examine the number of unique values in each column

unique_values = df.nunique()

unique_values.sort_values()
```

```python
# Identify columns with only one unique value

constant_columns = [
    column for column in df.columns
    if df[column].nunique() == 1
]

print("Constant Columns:")

print(constant_columns)
```

VISUALISATION

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Attrition"
)

plt.title("Distribution of Employee Attrition")

plt.xlabel("Attrition Status")

plt.ylabel("Number of Employees")

plt.savefig(
    "attrition_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

9.2 Employee age distribution

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Distribution of Employee Age")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.savefig(
    "age_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="Age"
)

plt.title("Employee Age by Attrition Status")

plt.xlabel("Attrition")

plt.ylabel("Age")

plt.show()
```
